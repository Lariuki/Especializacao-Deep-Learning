```python
import kagglehub
import os
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("munumbutt/amexfeather")

print("Path to dataset files:", path)
```

```python
os.listdir(path)
```

```python
train_path = os.path.join(path, "train_data.ftr")
train_path_f32  = os.path.join(path, "train_data_f32.ftr")

train_df = pd.read_feather(train_path)
train_df_f32  = pd.read_feather(train_path_f32)
```
